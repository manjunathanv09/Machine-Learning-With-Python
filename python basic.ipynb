```python
#Based on Percentage display result 85 to 100 is distinction,60 to 84 is first class,50 to 59 is second class,35 to 49 is pass,<35 is fail

per=float(input("Enter the percentage:"))
if(per>=85 and per<=100):
    print("Distinction")
elif(per>=60 and per<=84):
    print("first class")
elif(per>=50 and per<=59):
    print("second class")
elif(per>=35 and per<=49):
    print("Pass")
elif(per>=0 and per<=34):
    print("fail")
else:
    print("Invalid Input")
```

```python
#check the given inputs that they form which type of triangle Equilateral,isosceles,scalene

a=int(input("enter 1st side:"))
b=int(input("enter 2nd side:"))
c=int(input("enter 3rd side:"))
if(a+b<c or a+c<b or b+c<a):
    print("its not a triangle")
elif(a==b and b==c and c==a):
    print("Equilateral")
elif(a==b or b==c or c==a):
    print("Isosceles")
else:
    print("Scalene")
```

```python
#Electricity Bill Calculation , Scenario : an electricity company charges its customers as follows:
#First 100 units : $0.5 per unit
#Next 100 units (101-200): $0.75 per unit
#Above 200 units :$1 per unit

units = float(input("enter the Units"))
if(units>=100 and units<0):
    ta = units*0.5
elif(units>100 and units<=200):
    ta =(100*0.5)+((units-100)*0.75)
elif(units>200):
    ta =(100*0.5)+(100*0.75)((units-200)*1)
print("the total amount is:",ta)
```
