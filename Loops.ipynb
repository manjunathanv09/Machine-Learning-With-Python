```python
es = 0
os = 0
for i in range(1,21):
    if(i%2==0):
        es = es+ i
            #0+2
            #2+4=6
    else:
        os = os+i
             #0+1
             #1+3=4
print("the even sum is :",es)
print("the odd sum is :",os)
```

```python
pw =input("enter the password:")
uc = 0
lc = 0
sp = 0
nd = 0
if(len(pw)>=8):
    for i in pw:
        if i.isupper():
            uc+=1
        elif i.islower():
            lc+=1
        elif i.isdigit():
            nd+=1
        else:
            sp+=1
else:
    print("password length it should be more than 7 charcters")

if(uc>=1  and lc>=1 and nd>=1 and sp>=1):
    print("strong password")
else:
    print("weak password")
        
```

```python
num = int(input("enter the number:"))
fact = 1
for i in range(num,0,-1):
    fact = fact * i
          # 1    * 7
           #7    * 6
           #42   * 5
print("the factorial of given num is :",fact)
```

```python
# print the given num is prime number or not

num = int(input("enter the number:"))
count = 0
for i in range (2,num,1):
    if (num%i==0):
        count = count+1
if(count==0):
    print(num,"is a prime number")
else:
    print(num,"is not a prime number")
```
